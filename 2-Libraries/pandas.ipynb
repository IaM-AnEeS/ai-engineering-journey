##### Pandas is a fast, powerful, and flexible open-source Python library used for data analysis, manipulation, and cleaning. Created by Wes McKinney in 2008, its name is derived from the econometrics term "panel data" (multidimensional datasets involving observations over time) and serves as a play on "Python data analysis". It is built directly on top of NumPy, meaning it leverages high-performance arrays while adding labeled axes to make working with structured data highly intuitive

```python
import pandas as pd
```

```python
# Series datatype  1-Dimensional array-like object that can hold many data types, such as integers, strings, floating point numbers, Python objects, etc. It is similar to a column in a spreadsheet or a SQL table.

series = pd.Series(["Toyota", "Honda", "Ford", "Chevrolet", "Nissan"])
series
```

```python
colour = pd.Series(["Red", "Blue", "Green", "Yellow", "Black"])
colour
```

```python
# DataFrame datatype  2-Dimensional labeled data structure with columns of potentially different types. It is similar to a spreadsheet or SQL table, or a dict of Series objects.

car_data= pd.DataFrame({"Car Make" : series, "Colour" : colour})
car_data
```

### Manipulating Data


```python
car_sales["Make"]=car_sales["Make"].str.lower()
car_sales
```

```python
car_sales_missing = pd.read_csv("car-sales-missing-data.csv")
car_sales_missing
```

```python
car_sales_missing["Odometer"].mean()
```

```python
car_sales_missing["Odometer"].fillna(car_sales_missing["Odometer"].mean(), inplace=True) # if u dont write inplace=True, then it will return a new series with filled values but will not change the original series. If you write inplace=True, then it will change the original series and return None.
car_sales_missing
```

```python
car_sales_missing.dropna()
```

```python
car_sales_missing
```

```python
car_sales_missing.dropna(inplace=True) # will make changes in the original DataFrame and return None. If you dont write inplace=True, then it will return a new DataFrame with dropped rows but will not change the original DataFrame. If you write inplace=True, then it will change the original DataFrame and return None.
```

```python
car_sales_missing_dropped = car_sales_missing.dropna() # will return a new DataFrame with dropped rows but will not change the original DataFrame. If you write inplace=True, then it will change the original DataFrame and return None.
```

```python
## Adding Column from Series
seats_column = pd.Series([5, 5, 5, 5, 5])

# New Column called Seats
car_sales["Seats"] = seats_column
car_sales
```

```python
car_sales.fillna({"Seats":5},inplace=True)
car_sales
```

```python
## Column from python List
fuel_usage = [7.5, 8.0, 6.5, 7.0, 8.5,9.7,9.1,5.7, 6.3, 7.2]
car_sales["Fuel per 100KM"] = fuel_usage
car_sales
```

```python
car_sales["Total Fuel Used (L)"] = car_sales["Odometer (KM)"]/100 * car_sales["Fuel per 100KM"]
car_sales
```

```python
# Create a column from a single value

car_sales["Number of Wheels"] = 4
car_sales
```
