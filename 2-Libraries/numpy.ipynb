#### Dot Product

```python
np.random.seed(0)
mat1 = np.random.randint(10,size=(5,3))
mat2 = np.random.randint(10,size=(5,3))
```

```python
mat1
```

```python
mat2
```

```python
# Element-wise multiplication(Hadamard product)
mat1 * mat2
```

```python
# Dot product
np.dot(mat1,mat2 )
# numbers on the inside do not match that why its giving error.
# we must make number on the inside match what we can do is take the transpose of mat1 and then apply dot product.
```

```python
# Taking Transpose of mat1 
mat1 = mat1.T
```

```python
mat1.shape,mat2.shape
```

```python
mat3 = np.dot(mat1,mat2 )
mat3
```

```python
mat3.shape # Shape of the mat3 is the outside number of mat1 and mat2
```

### Dot product example (nut butter sales)

```python
np.random.seed(0)
# Numbers of jars sold
sales_amount = np.random.randint(20, size=(5,3))
sales_amount
```

```python
## Creating weekly sales Dataframe

weekly_sales = pd.DataFrame(sales_amount,index=["Mon","Tue","Wed","Thu","Fri"],columns=["Almond butter","Peanut butter", "Cashew buttter",])
weekly_sales
```

```python
# Creating price arrray

prices = np.array([10,8,12]) 
prices
```

```python
prices.shape 
# Important: In NumPy, a 1D array does not have rows and columns. It only has a length of 3.
# Thats why we change the shape in the below column to make it 2D with 1Row and 3Column to make it work with the Pandas.
```

```python
prices = prices.reshape(1,3)
prices
```

```python
# Creating butter prices DataFrame
butter_prices = pd.DataFrame(prices,index=["price"],columns=["Almond butter","Peanut butter", "Cashew buttter",])
butter_prices
```

```python
prices
```

```python
sales_amount
```

```python
prices.shape, sales_amount.shape
```

```python
# you can see above the inside of the shape number is not matching to make it match we will take the Transpose of sales_amount
sales_amount = sales_amount.T
sales_amount
```

```python
total_sales = prices.dot(sales_amount)
total_sales
```
