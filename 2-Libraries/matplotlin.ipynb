### Matplotlib example workflow

```python
# 0. import matplotlib and get it ready for plotting in jupyter

%matplotlib inline
import matplotlib.pyplot as plt

# 1. Prepare data
x = [ 1,2,3,4]
y = [11,22,33,44]

# 2. Setup plot
fig,ax = plt.subplots(figsize=(10,10)) # (width height)

# 3. Plot Data
ax.plot(x,y)

# 4. Customize plot
ax.set(title="Simple Plot",
       xlabel="x-axis",
       ylabel="y-axis")

# 5. save and show(you save the whole figure)
fig.savefig(r"D:\AI Learning\2-Libraries\my_plot.png")
```

#### Two options for subplots

```python
## There are two ways to make subplots
# Subplot option 1
fig,((ax1,ax2),(ax3,ax4)) = plt.subplots(nrows=2,
                                         ncols=2,
                                         figsize=(10,5))

# Plot to each different axis
ax1.plot(x,x/2);
ax2.scatter(np.random.random(10),np.random.random(10));
ax3.bar(nu_butter_prices.keys(),nu_butter_prices.values());
ax4.hist(x);
```

```python
# Subplots Option 2

fig, ax = plt.subplots(nrows=2,ncols=2,figsize=(10,5))

ax[0,0].plot(x,x/2);
ax[0,1].scatter(np.random.random(10),np.random.random(10));
ax[1,0].bar(nu_butter_prices.keys(),nu_butter_prices.values());
ax[1,1].hist(x);
```

### Which one should we choese/use?? (Pyplot vs OO matplotlib method )
*  When plotting something quickly, okay to use pyplot method
*  When plotting something more advanced, use the oo method

```python
heart_disease.head()
```

```python
over_50 = heart_disease[heart_disease["age"]>50]
over_50.head()
```

```python
## Pyplot method
over_50.plot(kind='scatter',
             x='age',
             y='chol',
             c='target');
```

```python
## OO method
fix,ax = plt.subplots(figsize=(10,6))
over_50.plot(kind='scatter',
             x='age',
             y='chol',
             c='target',
             ax=ax);
ax.set_xlim(45,100)
```

```python
over_50.target.values
```

```python
## OO method from scratch
fig,ax = plt.subplots(figsize=(10,6))

# plot the data
scatter = ax.scatter(x=over_50["age"],
                     y=over_50["chol"],
                     c=over_50["target"]);

## Customize
ax.set(title="Heart Disease and Cholesterol Levels",
       xlabel="Age",
       ylabel='Cholestrol')

## Add a legend
ax.legend(*scatter.legend_elements(),title="target");

## Add horizontal line 
ax.axhline(over_50["chol"].mean(),
           linestyle='--')
```

```python
over_50.head()
```

```python
## Subplot of chol, age, thalach
fig,(ax0,ax1) = plt.subplots(nrows=2,
                            ncols=1,
                            figsize=(10,10),
                            sharex=True);
# Add data to ax0
scatter = ax0.scatter(x=over_50["age"],
                      y=over_50["chol"],
                      c=over_50["target"]);
# Customize axo
ax0.set(title="Heart Disease and Cholestrol Levels",
        # xlabel="Age", Now we dont need to give x again we made sharex true above in plots making
        ylabel="Cholestrol");
## Add a legend
ax0.legend(*scatter.legend_elements(),title="Target")
## Add meanline
ax0.axhline(y=over_50["chol"].mean(),
            linestyle="--" );


## Add data to ax1
scatter = ax1.scatter(x=over_50["age"],
                      y=over_50["thalach"],
                      c=over_50["target"]);
# Customize axo
ax1.set( title="Heart Disease and Max Heart Rate",
         xlabel="Age", 
         ylabel="Thalach");
## Add a legend
ax1.legend(*scatter.legend_elements(),title="Target")
## Add meanline
ax1.axhline( y = over_50["thalach"].mean(),
             linestyle ="--" );


## Add a title to the figure
fig.suptitle("Heart Disease Analysis", fontsize=14, fontweight="bold");
```

```python
## Subplot of chol, age, thalach
fig,(ax0,ax1) = plt.subplots(nrows=2,
                            ncols=1,
                            figsize=(10,10),
                            sharex=True);
# Add data to ax0
scatter = ax0.scatter(x=over_50["age"],
                      y=over_50["chol"],
                      c=over_50["target"],
                      cmap="winter");
# Customize axo
ax0.set(title="Heart Disease and Cholestrol Levels",
        # xlabel="Age", Now we dont need to give x again we made sharex true above in plots making
        ylabel="Cholestrol");
## Add a legend
ax0.legend(*scatter.legend_elements(),title="Target")

# Change the x axis limits
ax0.set_xlim(50,80)
ax1.set_ylim(60,200)
## Add meanline
ax0.axhline(y=over_50["chol"].mean(),
            linestyle="--" );


## Add data to ax1
scatter = ax1.scatter(x=over_50["age"],
                      y=over_50["thalach"],
                      c=over_50["target"],
                      cmap="winter");
# Customize axo
ax1.set( title="Heart Disease and Max Heart Rate",
         xlabel="Age", 
         ylabel="Thalach");
## Add a legend
ax1.legend(*scatter.legend_elements(),title="Target")
## Add meanline
ax1.axhline( y = over_50["thalach"].mean(),
             linestyle ="--" );


## Add a title to the figure
fig.suptitle("Heart Disease Analysis", fontsize=14, fontweight="bold");
```
